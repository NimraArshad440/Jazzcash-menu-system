```python

code = input("Enter the code: ")

if code == "*786#":
    print("Jazzcash")
    print("1: Send Money")
    print("2: Pay Bills")
    print("3: Load & Bills")
    print("4: ReadyCash")
    print("5: Payments")

    choice = input("Select your option: ")

    if choice == "1":
        print("Send Money")
        print("1: Jazzcash Account")
        print("2: Bank Account")
        print("3: CNIC")

        sub_choice = input("Select Account Type: ")

        if sub_choice == "1":
            print("Jazzcash account selected")

            mobile = input("Enter Mobile number: ")
            print("Mobile number:", mobile)

            amount = input("Enter Amount: ")
            print("Amount:", amount)

            pin = input("Enter your PIN: ")

            if pin == "1234":
                print("Pin Verified")

                confirm = input("Do you want to confirm? (yes/no): ")

                if confirm.lower() == "yes":
                    print("Transaction Successful")
                else:
                    print("Transaction Cancelled")

            else:
                print("Incorrect PIN")

        elif sub_choice == "2":
            print("Bank Account selected")

        elif sub_choice == "3":
            print("CNIC selected")

        else:
            print("Invalid Account Choice")

    elif choice == "2":
        print("Pay Bills")

    elif choice == "3":
        print("Load & Bundles")

    elif choice == "4":
        print("ReadyCash")

    elif choice == "5":
        print("Payments")

    else:
        print("Invalid Option")

else:
    print("Invalid Code")
```
